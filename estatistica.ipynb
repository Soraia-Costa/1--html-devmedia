#Estatistica basica

```python
import numpy as np
```

```python
lista_idades = [26,30,32,22,26,35,40,20,43,31,23]
```

```python
sum(lista_idades)
```

```python
#contando a quantidade de itens na lista
len(lista_idades)
```

```python
#usando funcao soma
np.sum(lista_idades) / len(lista_idades)
```

```python
media = np.mean(lista_idades)
```

```python
#Usando media
media
```

```python
print('Media aritmetica:',media)
```

```python
#Colocando as idades na ordem:
lista_idades.sort()
lista_idades
```
